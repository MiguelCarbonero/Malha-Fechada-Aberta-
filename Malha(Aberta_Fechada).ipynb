```python
from tbcontrol.symbolic import routh
import sympy
```

```python
s = sympy.Symbol('s')
K = sympy.Symbol('K')
```

```python
C_s = K
P_s = (s+20)/(s*(s+2) * (s+3))
G_s = (C_s*P_s)
print ("Malha aberta G(s) =  {}".format(G_s))
```

```python
H_s = 1
G1_s = sympy.cancel((G_s)/(1+G_s*H_s))
print ("Malha fechada G1(s) =  {}".format(G1_s))
```

```python
num_G1_s, den_G1s_s = sympy.fraction(G1_s)
```

```python
tab_routh = routh(sympy.poly(den_G1s_s,s))
K_est = sympy.solve([e>0 for e in tab_routh[:,0]],K)
sympy.pprint("\Sistema estável ser: {}".format(K_est))
print("\nTabela de Routh: ")
routh(sympy.poly(den_G1s_s,s))
```
